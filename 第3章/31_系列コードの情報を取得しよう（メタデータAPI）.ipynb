## メタデータAPIから系列コードの情報を取得

```python
import requests

url = "https://www.stat-search.boj.or.jp/api/v1/getMetadata"

response = requests.get(url, params={"DB":"PR01"})

data = response.json()

series_list = [
    {
        "SERIES_CODE": item["SERIES_CODE"],
        "NAME_OF_TIME_SERIES_J": item["NAME_OF_TIME_SERIES_J"],
        "CATEGORY_J": item["CATEGORY_J"],
    }
    for item in data.get("RESULTSET", [])
    if item.get("SERIES_CODE")
]

series_list

# 必要であれば、以下のコードでCSVファイルとしてダウンロードする
import csv
from google.colab import files

csv_path = "series_codes.csv"
fieldnames = ["SERIES_CODE", "NAME_OF_TIME_SERIES_J", "CATEGORY_J"]

with open(csv_path, "w", newline="", encoding="utf-8-sig") as csv_file:
    writer = csv.DictWriter(csv_file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(series_list)

files.download(csv_path)
```
