## コードAPIから時系列統計データを取得

import requests

url = "https://www.stat-search.boj.or.jp/api/v1/getDataCode"

codes = [
    "PRCG20_2200010001",
    "PRCG20_2202010001",
    "PRCG20_2202110001",
    "PRCG20_2202210001",
    "PRCG20_2202310001",
]

response = requests.get(
    url,
    params={
        "DB": "PR01",
        "CODE": ",".join(codes),
        "FORMAT": "JSON",
        "LANG": "JP",
        "STARTDATE": "202504",
        "ENDDATE": "202603",
    },
)

response.json()
